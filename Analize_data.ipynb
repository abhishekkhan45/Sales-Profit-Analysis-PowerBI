```python
# read file data
df=pd.read_csv(r'D:\Data Analysis 1\Pythan\pandasBeeProject\Sampledata.xlsx.csv')
print(df.head())
```

```python
# check data
# check column 
print(df.columns)

#check datatype of column
print(df.dtypes)

```

```python
#check null value
print(df.isnull().sum())

#check duplicate row value
print(df.duplicated)
```

```python
#fill discount Band duplicate value
df["Discount Band"]=df["Discount Band"].fillna("No Discount")

# remmove duplicate row
df=df.drop_duplicates()
```

```python
#check duplicate row
print(df.duplicated().sum())
```

```python
#change date column datatpe
df["Date"]=pd.to_datetime(df["Date"])

print(df["Date"].dtypes)
```

```python
#check missing value
print(df.isnull().sum())
```

```python
#convert numeric column profit,unitsold, etc
numeric_column = [
    "Units Sold",
    "Manufacturing Price",
    "Sale Price",
    "Gross Sales",
    "Discounts",
    " Sales",
    "COGS",
    "Profit",
    "Month Number",
    "Year"
]

for i in numeric_column:
    df[i] = pd.to_numeric(df[i], errors="coerce")

print(df.dtypes)
```

```python
numeric_column=[" Sales","Profit","Gross Sales","Units Sold"

]
for i in numeric_column:
    df[i] = pd.to_numeric(df[i], errors="coerce").fillna(0).astype("int64")
```

```python
# Convert text columns to string
text_columns = [
    "Segment",
    "Country",
    "Product",
    "Discount Band",
    "Month Name"
]

for col in text_columns:
    df[col] = df[col].astype("string")

print(df.dtypes)   
```

```python
%pip install numpy matplotlib

import numpy as np
import matplotlib.pyplot as plt

print("NumPy version:", np.__version__)
print("Matplotlib version:", plt.matplotlib.__version__)
```

```python
#find total Profit
total_profit=df["Profit"].sum()
print("Tota; Profit",total_profit)
```

```python
#Year wise total profit
Year_profit=df.groupby("Year")["Profit"].sum().reset_index()
print(Year_profit)


# Visualization
plt.figure(figsize=(10, 5))

plt.bar(df["Year"],df["Profit"])

plt.xlabel("Year")
plt.ylabel("Total Profit")
plt.title("Year-wise Total Profit")

plt.xticks(df["Year"])
plt.show()
```

```python
#Discount By total profit
discount_profit= df[df["Profit"] > 0].groupby("Discount Band")["Profit"].sum().reset_index()
print(discount_profit)


```

```python
#Top 10 low products by profit
loss_products = (
    df[df["Profit"] > 0]
    .groupby("Product")["Profit"]
    .sum()
    .sort_values(ascending=True)
)

print(loss_products)
```

```python
#find Top 5 profit-making products — postive
top5_negative = (
    df[df["Profit"] > 0]
    .groupby("Product")["Profit"]
    .sum()
    .sort_values(ascending=True)
    .head(5)
)

print(top5_negative)



#Visualization

top5_negative.plot(kind="bar", figsize=(8,5))

plt.title("Top 5 profit-Making Products")
plt.xlabel("Product")
plt.ylabel("Total Loss")
plt.xticks(rotation=45)
plt.show()
```

```python
#find Top 5 negtive products 
top5_negative = (
    df[df["Profit"] < 0]
    .groupby("Product")["Profit"]
    .sum()
    .sort_values(ascending=True)
    .head(5)
)

print(top5_negative)



#Visualization

top5_negative.plot(kind="bar", figsize=(8,5))

plt.title("Top 5 negtive Products")
plt.xlabel("Product")
plt.ylabel("Total Loss")
plt.xticks(rotation=45)
plt.show()
```

```python
#highest unit sold by product
product_units = df.groupby("Product")["Units Sold"].sum().sort_values(ascending=False)

# Top 10 products
top10_units = product_units.head(10)

print(top10_units)



#Visualization — Top 10 Products by Units Sold

top10_units.plot(kind="bar", figsize=(10,5))

plt.title("Top 10 Products by Units Sold")
plt.xlabel("Product")
plt.ylabel("Total Units Sold")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#highest unit sold by product
product_units_low = df.groupby("Product")["Units Sold"].sum().sort_values(ascending=True)

# Top 10 products
top10_units_low = product_units_low.head(10)

print(top10_units_low)



#Visualization — lowest 10 Products by Units Sold

top10_units_low.plot(kind="bar", figsize=(10,5))

plt.title("lowest 10 Products by Units Sold")
plt.xlabel("Product")
plt.ylabel("Total Units Sold")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
